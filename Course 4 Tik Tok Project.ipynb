==> ENTER YOUR RESPONSE HERE (Double Click) There is a significant difference between the mean of the verfied users and un- verified users

Your goal in this step is to conduct a two-sample t-test. Recall the steps for conducting a hypothesis test:

State the null hypothesis and the alternative hypothesis
Choose a signficance level
Find the p-value
Reject or fail to reject the null hypothesis
==> ENTER YOUR NULL AND ALTERNATIVE HYPOTHESES HERE (Double Click) Ho= There is no difference in number of views of video posted by verified user and not verified users Ha= There is a difference in number of views of videos posted by verified and not verified users

```python
# Conduct a two-sample t-test to compare means
### YOUR CODE HERE ###
not_verified= data[data['verified_status']== 'not verified']['video_view_count']
verified= data[data['verified_status']== 'verified']['video_view_count']
```

```python
stats.ttest_ind(a= not_verified, b= verified, equal_var= False)
```

The p-VALUE IS 2.6088 is smaller than 5 therefore we reject the null hypothesis and concluse that There is a difference in number of views posted by verified user and number of views posted by un verified users
